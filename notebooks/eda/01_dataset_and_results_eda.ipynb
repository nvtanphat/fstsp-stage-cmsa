# FSTSP EDA
Notebook này chỉ dùng cho EDA (khám phá dữ liệu) và phân tích output. Thuật toán chính nằm trong `src/`.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parents[1] / "src"))
import pandas as pd
import matplotlib.pyplot as plt
from fstsp.data.generator import generate_uniform_instance
```

```python
inst = generate_uniform_instance(50, seed=42)
df = pd.DataFrame(inst.coords, columns=["x","y"])
df.describe()
```

```python
fig, ax = plt.subplots(figsize=(7,7))
ax.scatter(df.x, df.y)
ax.scatter([inst.depot_coord[0]],[inst.depot_coord[1]], marker="s", s=90)
ax.set_title("Generated FSTSP instance")
ax.set_aspect("equal")
plt.show()
```
